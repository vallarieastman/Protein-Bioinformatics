```python
# Load the dataset

import pandas as pd

df1 = pd.read_excel("Peptide Sequences.xlsx", skiprows=5)

# Clean up column names: remove leading/trailing spaces and line breaks
#df1.columns = df.columns.str.strip()
df1.columns = df1.columns.str.strip().str.replace('\n', ' ', regex=True)

# drop all "Unnamed" columns
df1 = df1.loc[:, ~df1.columns.str.contains("Unnamed")]

# Inspect columns to see what information the file has
print("Columns:", df1.columns.tolist())
df1.head()
```

```python
# Extract peptide sequences

# Replace column name if your file uses something slightly different
peptide_col = "Peptide Sequence"

peptides = df1[peptide_col]
print("Total peptide entries (including duplicates):", peptides.shape[0])
```

```python
#Remove duplicates & Count total peptide sequences

# Step 1: Extract the peptide sequence column
peptides = df1["Peptide Sequence"]

# Step 2: Remove duplicates
unique_peptides = peptides.drop_duplicates().reset_index(drop=True)

# Step 3: Print count
print("Unique peptide sequences:", unique_peptides.shape[0])
```

```python
# Count peptides per PTM category (PTMs are inside the peptide sequence)

import re

total_peptides = df1.shape[0]

def has_ph(seq):
    return "ph" in seq.lower()

def has_ac(seq):
    return "ac" in seq.lower()

def has_de(seq):
    return "de" in seq.lower()

df1['has_ph'] = df1['Peptide Sequence'].apply(has_ph)
df1['has_ac'] = df1['Peptide Sequence'].apply(has_ac)
df1['has_de'] = df1['Peptide Sequence'].apply(has_de)

summary = pd.DataFrame({
    "PTM Type": ["Phosphorylated", "Acetylated", "Deamidated"],
    "Count": [df1['has_ph'].sum(), df1['has_ac'].sum(), df1['has_de'].sum()]
})

summary["Percent (%)"] = (summary["Count"] / total_peptides * 100).round(2)

# Add total row
total_row = pd.DataFrame({
    "PTM Type": ["Total peptides"],
    "Count": [total_peptides],
    "Percent (%)": [100.0]
})

summary_with_total = pd.concat([summary, total_row], ignore_index=True)
summary_with_total
```

```python
# Load Dataset & Clean Columns

# Load Dataset 2
df2 = pd.read_excel("Regulated Peptides.xlsx", skiprows=7)

# Clean column names: strip spaces and line breaks
# df2.columns = df2.columns.str.strip()
df2.columns = df2.columns.str.strip().str.replace('\n', ' ', regex=True)

# drop all "Unnamed" columns
df2 = df2.loc[:, ~df2.columns.str.contains("Unnamed")]

print("Columns:", df2.columns.tolist())
df2.head()
```

```python
# Extract all UniProt IDs (AB column) and remove duplicates

# Try to detect a UniProt / Protein ID column automatically
candidate_cols = [c for c in df2.columns if "uniprot" in c.lower() or "protein id" in c.lower()]

print("Candidate protein ID columns:", candidate_cols)
```

```python
uniprot_col = candidate_cols[0]  # e.g. "Protein ID"

uniprot_ids = df2[uniprot_col]

print("Total IDs (with duplicates):", uniprot_ids.shape[0])
```

```python
# Remove Duplicates
unique_uniprot_ids = uniprot_ids.dropna().drop_duplicates().reset_index(drop=True)

print("Unique UniProt IDs:", unique_uniprot_ids.shape[0])
print("Total IDs (with duplicates & NAs):", uniprot_ids.shape[0])

# small DataFrame:
identified_proteins = pd.DataFrame({"UniProt ID": unique_uniprot_ids})
identified_proteins.head()
```

```python
import re
import pandas as pd

def canonical_peptide(seq):
    """
    Turn any peptide string with PTM annotations into plain AA sequence.
    Keeps only A–Z, throws away ph/ac/de, parentheses, etc.
    """
    if not isinstance(seq, str):
        return None
    seq = seq.upper()
    # remove explicit ptm tags like PH, AC, DE in df2 *after* uppercasing
    seq = re.sub(r"(PH|AC|DE)", "", seq)
    # now keep only A-Z
    return re.sub(r"[^A-Z]", "", seq)

# For Dataset 1
df1["pep_clean"] = df1["Peptide Sequence"].apply(canonical_peptide)

# For Dataset 2
df2["pep_clean"] = df2["Peptide sequence"].apply(canonical_peptide)

df1[["Peptide Sequence", "pep_clean"]].head()
df2[["Peptide sequence", "pep_clean"]].head()
```

```python
set1 = set(df1["pep_clean"].dropna().unique())
set2 = set(df2["pep_clean"].dropna().unique())

len(set1 & set2), len(set1), len(set2)
```

```python
pep_protein_map = (
    df2[["pep_clean", "Protein ID"]]
    .dropna(subset=["Protein ID"])
    .drop_duplicates(subset=["pep_clean"])
)

pep_protein_map.head()
```

```python
# Merge Dataset 1 + Dataset 2 by peptide sequence

df1_annotated = df1.merge(
    pep_protein_map,
    on="pep_clean",   # same column name in both now
    how="left"
)

df1_annotated.head()
```

```python
mapped = df1_annotated["Protein ID"].notna().sum()
total  = df1_annotated.shape[0]

print(f"Peptides with mapped UniProt ID: {mapped} / {total}")
```

```python
# Count PTMs by Type

# How many peptides in df1 have at least one of each PTM type?
ptm_counts_df1 = pd.DataFrame({
    "PTM type": ["Phosphorylated", "Acetylated", "Deamidated"],
    "Peptides with ≥1 site": [
        df1["has_ph"].sum(),
        df1["has_ac"].sum(),
        df1["has_de"].sum()
    ]
})

ptm_counts_df1

total_peptides_df1 = df1.shape[0]
ptm_counts_df1["Percent of all peptides (%)"] = (
    ptm_counts_df1["Peptides with ≥1 site"] / total_peptides_df1 * 100
).round(2)

ptm_counts_df1
```

```python
# Clean column names
df2.columns = df2.columns.str.strip()

# PTM columns
ptm_cols = ["Phospho", "Acetyl", "Glyco (put.)"]

# Convert PTM columns to numeric
df2[ptm_cols] = df2[ptm_cols].apply(pd.to_numeric, errors='coerce')

# Now compute site counts
ptm_site_counts_df2 = df2[ptm_cols].sum()
ptm_peptide_counts_df2 = (df2[ptm_cols] > 0).sum()

ptm_site_counts_df2, ptm_peptide_counts_df2
```

```python
# Determine how many have multiple PTMs

# Number of PTM types on each peptide (0–3)
df1["n_ptm_types"] = df1[["has_ph", "has_ac", "has_de"]].sum(axis=1)

n_multi_df1 = (df1["n_ptm_types"] >= 2).sum()
n_any_ptm_df1 = (df1["n_ptm_types"] >= 1).sum()

print("Peptides with ≥1 PTM type:", n_any_ptm_df1)
print("Peptides with ≥2 PTM types:", n_multi_df1)

multi_summary_df1 = df1["n_ptm_types"].value_counts().sort_index()
multi_summary_df1

# True if peptide has at least one site of each type
df2["has_ph"] = df2["Phospho"] > 0
df2["has_ac"] = df2["Acetyl"] > 0
df2["has_de"] = df2["Glyco (put.)"] > 0

# Number of PTM types per regulated peptide
df2["n_ptm_types"] = df2[["has_ph", "has_ac", "has_de"]].sum(axis=1)

n_multi_df2 = (df2["n_ptm_types"] >= 2).sum()
n_any_ptm_df2 = (df2["n_ptm_types"] >= 1).sum()

print("Regulated peptides with ≥1 PTM type:", n_any_ptm_df2)
print("Regulated peptides with ≥2 PTM types:", n_multi_df2)

df2["n_ptm_types"].value_counts().sort_index()
```

```python
# Compare regulated vs non-regulated PTMs

# Set of regulated canonical sequences
regulated_set = set(df2["pep_clean"].dropna())

# Boolean flag on df1
df1["is_regulated"] = df1["pep_clean"].isin(regulated_set)

df1["is_regulated"].value_counts()

def summarize_ptms(group):
    return pd.Series({
        "N peptides": group.shape[0],
        "N ph peptides": group["has_ph"].sum(),
        "N ac peptides": group["has_ac"].sum(),
        "N de peptides": group["has_de"].sum()
    })

#ptm_reg_summary = df1.groupby("is_regulated").apply(summarize_ptms)
ptm_reg_summary = df1.groupby("is_regulated", group_keys=True).apply(
    summarize_ptms,
    include_groups=False
)

# Make the index human-readable
ptm_reg_summary.index = ["Non-regulated", "Regulated"]

# Add percentages within each group
for col in ["N ph peptides", "N ac peptides", "N de peptides"]:
    ptm_reg_summary[col + " (%)"] = (
        ptm_reg_summary[col] / ptm_reg_summary["N peptides"] * 100
    ).round(2)

ptm_reg_summary
```

```python
# Creates a protein-level summary to annotate

# Mapped peptides only
mapped = df1_annotated[df1_annotated["Protein ID"].notna()].copy()

# Aggregate PTM info to the protein level
protein_summary = mapped.groupby("Protein ID").agg(
    N_peptides    = ("pep_clean", "nunique"),
    N_ph_pep      = ("has_ph", "sum"),
    N_ac_pep      = ("has_ac", "sum"),
    N_de_pep      = ("has_de", "sum")
)

# Boolean: does this protein ever appear with each PTM type?
protein_summary["has_ph"] = protein_summary["N_ph_pep"] > 0
protein_summary["has_ac"] = protein_summary["N_ac_pep"] > 0
protein_summary["has_de"] = protein_summary["N_de_pep"] > 0

protein_summary.head()
```

```python
# Export UniProt IDs for batch annotation

protein_ids = protein_summary.index.to_series().reset_index(drop=True)
protein_ids.to_csv("protein_ids_for_uniprot.txt", index=False, header=False)
```

```python
# Use earlier PTM flags to add categorical annotations

def ptm_category(row):
    types = []
    if row["has_ph"]:
        types.append("phospho")
    if row["has_ac"]:
        types.append("acetyl")
    if row["has_de"]:
        types.append("deamidation")
    if not types:
        return "no PTM"
    if len(types) == 1:
        return types[0]
    return "multi-PTM (" + "+".join(types) + ")"

protein_annotated["PTM_category"] = protein_annotated.apply(ptm_category, axis=1)
protein_annotated["PTM_category"].value_counts()
```

```python
# Export list for Reactome (Requires one protein per line)

# Start from your merged peptide table
mapped = df1_annotated[df1_annotated["Protein ID"].notna()].copy()

# Build a clean Series of UniProt IDs
prot_ids = (
    mapped["Protein ID"]
    .astype(str)                        # make sure it's string
    .str.replace(r"\s+", "", regex=True)  # remove *all* spaces inside
    .str.split(";")                     # split on semicolons
    .explode()                          # turn each element into its own row
)

# Drop empties
prot_ids = prot_ids[prot_ids != ""].dropna()

# Deduplicate
prot_ids = prot_ids.drop_duplicates().reset_index(drop=True)

print("Number of unique IDs:", prot_ids.shape[0])
prot_ids.head(30)
```

Across the developmental time points, the peptides showing the largest iTRAQ ratio changes mapped predominantly to proteins associated with mitochondrial metabolism, chromatin regulation, and cytoskeletal remodeling.     <b>2. Are extensive PTM modification (i.e. number of PTMs) and PTM regulation correlated? Do they appear to target the same areas of proteins?</b>

There was no strict correlation between the number of PTMs on a peptide and the likelihood of developmental regulation; however, regulated PTMs tended to occur within the same functional regions repeatedly targeted during development. Phosphorylated peptides were overrepresented in proteins belonging to mitochondrial pathways, chromatin-modifying complexes (HATs/HDACs, PRC2, MLL3/4), and metabolic regulators, indicating that specific residues within regulatory domains act as PTM “hotspots.” These hotspots often contained multiple PTMs of the same or different types, suggesting coordinated modification rather than random decoration. In contrast, peptides with many PTMs but no regulation typically mapped to proteins outside major developmental pathways. Overall, PTM density alone did not predict regulation, but regulated PTMs clearly clustered within conserved domains participating in mitochondrial remodeling, chromatin regulation, and metabolic transitions, showing that PTM regulation is region-specific rather than globally distributed.

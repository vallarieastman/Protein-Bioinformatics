   <b> IUPAC Definition: </b><br>
The IUPAC definition of a dihedral angle comes directly from how the potential V(r) depends on four atoms’ positions. The torsion angle is formed by four connected atoms A–B–C–D (the angle between the planes A–B–C and B–C–D) (IUPAC, 2008). It is measured in degrees (-180° and 180°). Atoms A-B-C-D can be more specifically defined as C–N–Cα–C (phi) and N–Cα–C–N (psi). For proteins, the backbone torsion angles are denoted as: <br>

Φ (phi): C(i−1)–N(i)–Cα(i)–C(i), rotates at the the N–Cα bond<br>
Ψ (psi): N(i)–Cα(i)–C(i)–N(i+1), rotates at the Cα–C’ bond<br>
Ω (omega):  Cα(i) – C(i) – N(i+1) – Cα(i+1), rotates at the C’-N bond<br><br>

<b>PDB File Coordinates: </b><br>
The positions of backbone atoms (N, Cα, C, O) in a PDB file define the 3D geometry of the protein. From these coordinates, any geometric property, including bond lengths, bond angles, and dihedral angles (φ, ψ), can be directly calculated. The PDB file coordinates are listed under the “atom_site” loop. To calculate a dihedral angle from four consecutive atoms (labeled A, B, C, and D), three bond vectors are constructed: v₁ = A − B, v₂ = B − C, and v₃ = C − D. The orientation of the two planes formed by these atoms is determined by finding their respective normal vectors using cross products: m = v₁ × v₂ for the plane defined by A–B–C, and n = v₂ × v₃ for the plane defined by B–C–D. The angle between these two planes, which defines the dihedral angle, is obtained using the dot product formula cos(θ) = (m ⋅ n) / (|m||n|), where m ⋅ n is the dot product of the normal vectors and |m|, |n| are their magnitudes. Finally, the dihedral angle θ is computed as θ = arccos[(m ⋅ n) / (|m||n|)], yielding the torsion angle that characterizes the rotation around the central bond connecting atoms B and C.<br><br>

    The strategy for assigning secondary structure from backbone coordinates involves calculating the dihedral angles phi (φ) and psi (ψ) for each residue using the atomic coordinates. These angles are derived from four consecutive backbone atoms: C′(i–1)–N(i)–Cα(i)–C′(i) for φ and N(i)–Cα(i)–C′(i)–N(i+1) for ψ.  φ describes rotation about the N–Cα bond, and ψ describes rotation about the Cα–C′ bond.α-helices typically occur near φ ≈ –60° and ψ ≈ –45°, β-strands near φ ≈ –120° and ψ ≈ 120°, and all other conformations, such as loops or turns, fall outside these ranges ( Rigden, 2017). Each residue is classified as α (helical), β (extended), or other (irregular) based on these boundaries. This φ/ψ-based approach is the foundation of DSSP.<br><br>
Each (φ, ψ) pair can be compared to known regions on the Ramachandran map that correspond to specific secondary structures. Steric constraints between atoms limit these rotations to specific combinations, producing distinct clusters for common secondary structures.To read a Ramachandran plot, locate each residue’s φ (x-axis) and ψ (y-axis) angles; points clustered in the lower left correspond to α-helical conformations, those in the upper left to β-strand conformations, and scattered points elsewhere indicate loops or irregular regions. The Ramachandran plot provides a direct geometric method for identifying and classifying backbone conformations into secondary structure types ( Rigden, 2017).<br><br>
    The most widely used programs for predicting secondary structure directly from dihedral (φ/ψ) coordinates are DSSP, STRIDE, and PROMOTIF. DSSP (Dictionary of Secondary Structure of Proteins) computes exact geometry derived from φ/ψ angles. STRIDE uses both hydrogen-bond energy and φ/ψ angle statistics to improve agreement with experimental data (Wu et al., 2017). PROMOTIF extends DSSP by identifying higher-order motifs such as β-hairpins, sheets, and helix–turn–helix units. Beyond these rule-based methods, more advanced strategies incorporate consensus, contact-based, and machine-learning approaches. Consensus methods combine multiple assignment algorithms to reduce noise and produce a more stable consensus classification for ambiguous regions. Machine-learning approaches use neural networks or support vector machines trained on large structural datasets to recognize secondary structure from backbone features, including φ/ψ, residue type, and local environment (Haspel et al., 2022). <br><br>
My approach uses DSSP and STRIDE, which add hydrogen-bond and geometric criteria for more precise structure assignment. DSSP identifies hydrogen bonds and assigns structural states according to consistent φ/ψ and hydrogen-bonding patterns. STRIDE refines this further using empirical angle cutoffs and energy calculations, and provides validation.

# b. Compute Phi and Psi Angles 

```python
# Computes phi and psi, then prints it to a table with formula explainations 
import mdtraj as md
import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import HTML, display
from Bio.PDB import PDBParser

# --- paths ---
BASE_DIR = Path("/Users/val/Projects/Protein-Bioinformatics/midterm")
clean_pdb = BASE_DIR / "7EU8_chainA_clean.pdb"
out_csv = BASE_DIR / "7EU8_chainA_phi_psi.csv"

# --- load structure with MDTraj ---
traj = md.load_pdb(str(clean_pdb))
top = traj.topology

# --- compute backbone torsions ---
phi_indices, phi_values = md.compute_phi(traj)   # radians
psi_indices, psi_values = md.compute_psi(traj)   # radians

phi_rad = phi_values[0]        # radians
psi_rad = psi_values[0]
phi_deg = np.degrees(phi_rad)  # degrees
psi_deg = np.degrees(psi_rad)

# --- helper to build human-readable block for each dihedral ---
def explained_dihedral(atom_indices, angle_rad, angle_deg):
    """
    Returns HTML block:

    C-28 = 241 | N-29 = 245 | CA-29 = 247 | C-29 = 250
    θ = dihedral(241, 245, 247, 250) using θ = atan2(b₂ᵤ · (n₁ × n₂), n₁ · n₂) = -1.08 rad
    (-1.08 × 180/π = -62.14°)
    """
    if atom_indices is None or np.isnan(angle_rad) or np.isnan(angle_deg):
        return ""

    atoms = [top.atom(int(a)) for a in atom_indices]

    # Line 1: map atom name/res -> PDB serial
    labeled_parts = []
    serial_ids = []
    for a in atoms:
        labeled_parts.append(f"{a.name}-{a.residue.resSeq} = {a.serial}")
        serial_ids.append(str(a.serial))
    line1 = " | ".join(labeled_parts)

    # Line 2: state the dihedral definition with atan2 reference and radian value
    serial_list = ", ".join(serial_ids)
    line2 = (
        f"θ = dihedral({serial_list}) using θ = atan2(b₂ᵤ · (n₁ × n₂), n₁ · n₂) "
        f"= {angle_rad:.2f} rad"
    )

    # Line 3: explicit rad→deg conversion
    line3 = f"({angle_rad:.2f} × 180/π = {angle_deg:.2f}°)"

    return f"{line1}<br>{line2}<br>{line3}"

# --- build table rows ---
rows = []
for i in range(len(phi_indices)):
    phi_quad = phi_indices[i]

    # residue for row = residue owning the N atom (2nd atom of φ quadruplet)
    res_for_row = top.atom(int(phi_quad[1])).residue
    resid = res_for_row.resSeq
    resname = res_for_row.name

    # φ angles
    this_phi_rad = float(phi_rad[i]) if i < len(phi_rad) else np.nan
    this_phi_deg = float(phi_deg[i]) if i < len(phi_deg) else np.nan

    # ψ angles (ψ may be undefined at C-terminus)
    if i < len(psi_indices):
        psi_quad = psi_indices[i]
        this_psi_rad = float(psi_rad[i]) if i < len(psi_rad) else np.nan
        this_psi_deg = float(psi_deg[i]) if i < len(psi_deg) else np.nan
    else:
        psi_quad = None
        this_psi_rad = np.nan
        this_psi_deg = np.nan

    # human-readable blocks
    phi_block = explained_dihedral(phi_quad, this_phi_rad, this_phi_deg)
    psi_block = explained_dihedral(psi_quad, this_psi_rad, this_psi_deg)

    rows.append({
        "pdb_resid": resid,
        "resname": resname,
        "phi_explained": phi_block,
        "psi_explained": psi_block,
        "phi_deg": round(this_phi_deg, 2) if not np.isnan(this_phi_deg) else np.nan,
        "psi_deg": round(this_psi_deg, 2) if not np.isnan(this_psi_deg) else np.nan,
    })

df = pd.DataFrame(rows)

# --- save CSV ---
df.to_csv(out_csv, index=False)
print(f"[info] wrote {out_csv}")


# --- BioPython parse of same file ---
parser = PDBParser(QUIET=True)
structure = parser.get_structure("7EU8_chainA", str(clean_pdb))

chains = [
    (
        chain.id,
        len([res for res in chain.get_residues() if res.id[0] == " "])
    )
    for model in structure
    for chain in model
]
print("Chains and (# aa residues):", chains)

first10 = []
for model in structure:
    for chain in model:
        if chain.id == "A":
            for res in chain.get_residues():
                if res.id[0] == " ":  # standard residue only
                    first10.append((res.id[1], res.resname))
                if len(first10) >= 10:
                    break
print("First visible residues in chain A:", first10)

# --- scrollable light-mode display ---
scroll_html = df.to_html(index=False, escape=False)
display(HTML(f"""
<div style="
    max-height:400px;
    overflow-y:auto;
    border:1px solid #ccc;
    border-radius:8px;
    padding:10px;
    font-family: ui-monospace, Menlo, Consolas, monospace;
    font-size:13px;
    background:#ffffff;
    color:#222;
    box-shadow:0 1px 4px rgba(0,0,0,0.08);
    line-height:1.4;
    white-space:nowrap;
">
{scroll_html}
</div>
"""))
```

# c. Using STRIDE for Validation

```python
# STRIDE data taken from web-based STRIDE application; this code cleans the file and prints
import re
import pandas as pd
from pathlib import Path
from IPython.display import HTML, display  # needed for scrollable preview

stride_path = Path("/Users/val/Projects/Protein-Bioinformatics/midterm/7EU8_STRIDE.txt")

rows = []
num_re = re.compile(r"[-+]?\d*\.\d+|\d+")

with open(stride_path, "r") as fh:
    for line in fh:
        # STRIDE secondary structure assignments are on lines that start with "ASG"
        if not line.startswith("ASG"):
            continue

        # Parse using fixed-width substrings (STRIDE format)
        resname = line[4:7].strip()          # residue name, e.g. ASN
        chain   = line[8:9].strip()          # chain ID, e.g. A
        pdb_res = line[10:15].strip()        # PDB residue number, e.g. 28
        seq_idx = line[16:21].strip()        # sequential index in chain
        ss_code = line[24:25].strip()        # 1-letter SS code (H,E,C,…)
        ss_full = line[27:43].strip()        # full SS label (AlphaHelix, Coil, etc.)

        phi_txt = line[44:53].strip()        # phi in degrees
        psi_txt = line[54:63].strip()        # psi in degrees
        area_txt= line[64:].strip()          # solvent exposure-ish

        # helper to robustly grab first number from the field
        def safe_float(x):
            m = num_re.findall(x)
            return float(m[0]) if m else None

        rows.append({
            "seq_idx": int(seq_idx),
            "pdb_resid": int(pdb_res),
            "resname": resname,
            "chain": chain,
            "ss_code_stride": ss_code,
            "ss_full_stride": ss_full,
            "phi_deg_stride": safe_float(phi_txt),
            "psi_deg_stride": safe_float(psi_txt),
            "accessibility": safe_float(area_txt),
        })

# turn into DataFrame and sort by PDB residue number
df_stride = (
    pd.DataFrame(rows)
    .sort_values("pdb_resid")
    .reset_index(drop=True)
)

# save full table to CSV
out_csv = Path("/Users/val/Projects/Protein-Bioinformatics/midterm/7EU8_chainA_STRIDE_per_residue.csv")
df_stride.to_csv(out_csv, index=False)
print(f"[info] wrote {out_csv} ({len(df_stride)} residues)")

# scrollable, light-mode preview of first 50 rows
display(HTML(f"""
<div style="
    max-height:400px;
    overflow-y:auto;
    border:1px solid #ccc;
    border-radius:8px;
    padding:10px;
    font-family: ui-monospace, Menlo, Consolas, monospace;
    font-size:12px;
    background:#fff;
    color:#222;
    line-height:1.4;
    white-space:nowrap;">
{df_stride.head(50).to_html(index=False)}
</div>
"""))
```

```python
# STRIDE Secondary Structure Prediction
helix_res = df_stride.loc[df_stride["ss_code_stride"] == "H", "pdb_resid"].tolist()
sheet_res = df_stride.loc[df_stride["ss_code_stride"] == "E", "pdb_resid"].tolist()
coil_res  = df_stride.loc[~df_stride["ss_code_stride"].isin(["H","E"]), "pdb_resid"].tolist()

def ranges(nums):
    # turn [36,37,38,39,41,42] -> ['36-39','41-42']
    if not nums:
        return []
    nums = sorted(nums)
    start = prev = nums[0]
    out = []
    for n in nums[1:]:
        if n == prev + 1:
            prev = n
        else:
            out.append(f"{start}-{prev}" if start != prev else f"{start}")
            start = prev = n
    out.append(f"{start}-{prev}" if start != prev else f"{start}")
    return out

helix_ranges = ranges(helix_res)
sheet_ranges = ranges(sheet_res)
coil_ranges  = ranges(coil_res)

print("Helix ranges:", helix_ranges)
print("Sheet ranges:", sheet_ranges)
print("Coil ranges:", coil_ranges)
```

```python
# DSSP data taken from web-based DSSP application; this parsers the .dssp file
import re
import pandas as pd
from pathlib import Path
import math

float_re = re.compile(r"[-+]?\d+\.\d+|[-+]?\d+(?=\s)")  
# explanation:
# - match signed floats like -65.4
# - OR signed ints like 360 (but avoid gluing trailing commas etc.)

def parse_dssp_legacy_regex(dssp_path):
    rows = []
    in_table = False

    with open(dssp_path, "r") as fh:
        for raw in fh:
            line = raw.rstrip("\n")

            # detect header boundary
            if line.lstrip().startswith("#  RESIDUE AA STRUCTURE"):
                in_table = True
                continue
            if not in_table:
                continue

            # skip non-residue lines
            if not re.match(r"\s*\d+", line):
                continue

            # --- fixed-width "front" ---
            dssp_idx_str  = line[0:5].strip()    # DSSP serial
            pdb_resid_str = line[5:10].strip()   # PDB residue number
            chain_id      = line[10:12].strip()  # chain
            aa            = line[12:15].strip()  # amino acid
            struct_block  = line[16:25]          # DSSP structure descriptor
            acc_str       = line[34:38].strip()  # Accessible surface area

            if aa == "!":    # chain break marker
                continue

            # convert indices
            try:
                dssp_idx = int(dssp_idx_str)
            except ValueError:
                continue

            try:
                pdb_resid = int(pdb_resid_str)
            except ValueError:
                continue

            sb_clean = struct_block.strip()
            primary_code = sb_clean[0] if sb_clean else ""

            # accessible surface area
            try:
                acc_val = float(acc_str)
            except ValueError:
                acc_val = math.nan

            # --- tail numeric parse using regex ---
            tail_txt = line[38:]
            # grab ALL numbers in order, including glued negatives
            nums = float_re.findall(tail_txt)

            # convert all to float
            try:
                nums = [float(x) for x in nums]
            except ValueError:
                nums = []

            # We expect ... TCO, KAPPA, ALPHA, PHI, PSI, Xca, Yca, Zca
            phi_val = math.nan
            psi_val = math.nan
            if len(nums) >= 5:
                phi_val = nums[-5]
                psi_val = nums[-4]

            rows.append({
                "dssp_idx": dssp_idx,
                "pdb_resid": pdb_resid,
                "chain": chain_id,
                "aa": aa,
                "dssp_code": primary_code,
                "dssp_block": sb_clean,
                "acc": acc_val,
                "phi_deg_dssp": phi_val,
                "psi_deg_dssp": psi_val,
            })

    return pd.DataFrame(rows)

# Run it
dssp_file = Path("/Users/val/Projects/Protein-Bioinformatics/midterm/7EU8_DSSPLegacy.dssp")
df_dssp = parse_dssp_legacy_regex(dssp_file)

print("Parsed rows:", len(df_dssp))
display(df_dssp.head(15))
```

```python
# Ramachandran colored by DSSP class
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# assume `merged` DataFrame is already in memory from the previous step
# and has columns:
# phi_mdtraj_deg, psi_mdtraj_deg, ss_stride, ss_dssp

def classify_stride(ss):
    """
    Collapse STRIDE's ss label into Helix / Sheet / Coil
    e.g. 'AlphaHelix', '310Helix', etc → Helix
         'Strand', 'Bridge' → Sheet
         else → Coil
    """
    if pd.isna(ss):
        return "Coil"
    s = ss.lower()
    if "helix" in s:
        return "Helix"
    if "strand" in s or "sheet" in s or "bridge" in s:
        return "Sheet"
    return "Coil"

def classify_dssp(code):
    """
    DSSP 1-letter code:
    H = alpha helix
    G = 3-10 helix
    I = pi helix
    E = extended strand
    B = beta bridge
    T, S, ' ' = turn/bend/loop/etc.
    """
    if pd.isna(code) or code.strip() == "":
        return "Coil"
    c = code.strip()[0]
    if c in ["H", "G", "I"]:
        return "Helix"
    if c in ["E", "B"]:
        return "Sheet"
    return "Coil"

plot_df = merged.copy()

plot_df["ss3_stride"] = plot_df["ss_stride"].apply(classify_stride)
plot_df["ss3_dssp"]   = plot_df["ss_dssp"].apply(classify_dssp)

# We'll also drop rows that don't have angles
plot_df = plot_df[
    ~plot_df["phi_mdtraj_deg"].isna() &
    ~plot_df["psi_mdtraj_deg"].isna()
].reset_index(drop=True)

# helper to make the plot
def rama_plot(df, ss_col, title):
    fig, ax = plt.subplots(figsize=(5,5))
    
    # split by class to get per-class scatter (easier legend)
    for ss_class in ["Helix", "Sheet", "Coil"]:
        sub = df[df[ss_col] == ss_class]
        ax.scatter(
            sub["phi_mdtraj_deg"],
            sub["psi_mdtraj_deg"],
            s=20,
            alpha=0.6,
            label=ss_class,
            edgecolors='black',
            linewidths=0.3
        )
    
    ax.set_xlabel("phi (°)")
    ax.set_ylabel("psi (°)")
    ax.set_title(title)
    ax.set_xlim(-180, 180)
    ax.set_ylim(-180, 180)
    ax.set_xticks(np.arange(-180, 181, 60))
    ax.set_yticks(np.arange(-180, 181, 60))
    ax.set_aspect('equal', 'box')
    ax.legend(frameon=False, loc="best")
    plt.show()

# 1) Ramachandran colored by STRIDE
rama_plot(plot_df, "ss3_stride", "Ramachandran: STRIDE secondary structure")

# 2) Ramachandran colored by DSSP
rama_plot(plot_df, "ss3_dssp", "Ramachandran: DSSP secondary structure")
```

```python
#Calculates Contact Order
from pathlib import Path
import mdtraj as md
import numpy as np

model_path  = Path("/Users/val/Projects/Protein-Bioinformatics/midterm/model1.pdb")
naive_path  = Path("/Users/val/Projects/Protein-Bioinformatics/midterm/7EU8_chainA.pdb")
mutant_path = Path("/Users/val/Projects/Protein-Bioinformatics/midterm/7EOQ_chainB_clean.pdb")  # we'll fix this later

def load_structure_any(path: Path):
    """
    Load structure with MDTraj.
    Uses md.load_pdb for .pdb and md.load for .cif/.mmcif.
    """
    suffix = path.suffix.lower()
    if suffix == ".pdb":
        return md.load_pdb(str(path))
    elif suffix in [".cif", ".mmcif"]:
        return md.load(str(path))
    else:
        raise ValueError(f"Unsupported file format: {suffix}")

def slice_chain(traj, desired_chain_id=None):
    """
    Return a trajectory containing only atoms from desired_chain_id.
    If desired_chain_id is None, return full traj.
    """
    if desired_chain_id is None:
        return traj

    top = traj.topology
    atom_indices = [
        atom.index
        for atom in top.atoms
        if atom.residue.chain.chain_id == desired_chain_id
    ]

    if len(atom_indices) == 0:
        # Chain not found, just return original so we don't die
        return traj

    return traj.atom_slice(atom_indices)

def contact_order_from_traj(traj, cutoff=0.8):
    """
    Compute contact order for a (possibly single-chain) traj.
    cutoff in nm (0.8 nm ~ 8 Å).
    """
    top = traj.topology

    # Get all Cα atoms in order
    ca_atoms = [atom.index for atom in top.atoms if atom.name == "CA"]
    if len(ca_atoms) < 2:
        raise ValueError("Not enough Cα atoms to compute contact map.")

    coords = traj.xyz[0, ca_atoms, :]  # (Nres, 3)

    # Map each CA atom to a residue index local to this traj
    # Build a "res_idx_in_chain" array that's 0..L-1 in order of first appearance
    residues_seen = []
    res_map = {}
    ca_res_ids = []
    for atom_idx in ca_atoms:
        res = top.atom(atom_idx).residue
        if res not in residues_seen:
            res_map[res] = len(residues_seen)
            residues_seen.append(res)
        ca_res_ids.append(res_map[res])
    ca_res_ids = np.array(ca_res_ids)  # length Nres
    L = len(residues_seen)

    # Pairwise distances between Cα atoms (nm)
    diff = coords[:, None, :] - coords[None, :, :]
    dists = np.sqrt(np.sum(diff**2, axis=-1))

    # Find contacting pairs under cutoff but not self
    contacts = np.argwhere((dists < cutoff) & (dists > 0))
    n_contacts = len(contacts)
    if n_contacts == 0:
        return np.nan, 0, L

    # Sequence separation |i-j|
    seq_sep = np.abs(ca_res_ids[contacts[:,0]] - ca_res_ids[contacts[:,1]])

    # Contact order formula
    co = np.sum(seq_sep) / (n_contacts * L)

    return co, n_contacts, L

def analyze_structure(label, path, chain_id=None):
    if not path.exists():
        print(f"{label:<35} | ERROR: file not found -> {path}")
        return

    try:
        traj_full = load_structure_any(path)
    except Exception as e:
        print(f"{label:<35} | ERROR loading {path.name}: {e}")
        return

    try:
        traj_chain = slice_chain(traj_full, desired_chain_id=chain_id)
        co, n_contacts, L = contact_order_from_traj(traj_chain)
        print(f"{label:<35} | Chain={chain_id or 'ALL'} | CO={co:.4f} | Contacts={n_contacts} | Residues={L}")
    except Exception as e:
        print(f"{label:<35} | ERROR processing {path.name}: {e}")


# --- Run it ---
# Try chain 'A' for first two because they're NR1 chain A PDBs
analyze_structure("SWISS-MODEL (WT-on-7EOQ)", model_path, chain_id="A")
analyze_structure("Naive 7EU8",                 naive_path, chain_id="A")

# 7EOQ currently mmCIF and causing parser grief; we'll handle conversion below
analyze_structure("Functional Mutant 7EOQ",     mutant_path, chain_id="B")
```

<p style="color: #34495e; font-family: 'Arial', sans-serif; font-size: 15px; margin-top: 18px;">
    Root Mean Square Deviation (RMSD) quantifies the average atomic displacement between two aligned structures. It’s computed after superimposing the structures to minimize overall distance. RMSD is computed by aligning the Cα atoms (backbone atoms) for two structures, superimposing the structures by translating (shifting) and rotating along the x, y, z axes to minimize the distances between corresponding atoms, and computing the Euclidean distance between corresponding atoms. Therefore, the shorter the distance, the smaller the RMSD, the more similar the structures. <br><br>
The Global Distance Test - Total Score (GDT_TS) measures how many residues in a predicted model are close to the same residues in the experimental structure, within certain distance cutoffs. GDT_TS represents the percentage of residues correctly positioned. Its calculated by selecting the Cα atoms in each structure, translating (shifting) and rotating along the x, y, z axes to minimize the distances between corresponding atoms, calculating the pairwise distances, count residues within each cutoff (e.g. the number of residues within 1 Å, 2 Å, etc.), and finally calculating the average percentage of residues in close proximity. Therefore, the higher the percentage, the more closely matched the structures.<br><br>
The GDT_TS score is used as the primary accuracy metric in CASP (Senior et al., 2019). RMSD gives equal weight to every atom, so a few poorly placed residues can inflate the RMSD. GDT_TS only considers whether residues fall within several distance cutoffs (1, 2, 4, 8 Å), so a few large deviations don’t significantly alter the score. This also means GDT_TS better captures the global topology, even if local regions are slightly misaligned, whereas local deviations alter RMSD even when the fold is functionally accurate. GDT_TS is given as a percentage of correctly placed residues, which makes it comparable across targets. <br><br>
*Additionally, the TM-Score quantifies the overall structural similarity between two protein structures, ranging from 0 to 1, where 1 indicates a perfect match. Just as the previous two measures, TM-Score also identifies aligned residues, superimposes the two structures, and calculates the pairwise distances, however, in its final calculation it uses weighted contribution of each residue and then normalizes the score. <br><br>

<h4 style="color: #7f8c8d; font-family: 'Georgia', serif; font-weight: normal; font-size: 20px; margin-top: 0;">
    Secondary Structure Motifs
  </h4>

  <p style="color: #34495e; font-family: 'Arial', sans-serif; font-size: 15px; margin-top: 18px;">
    The structure of a beta propeller domain in complex with peptide S (PDB: 3SOV) corresponds to a six-bladed beta-propeller. Each blade in a beta-propeller typically has four β-strands and the six blades together form a funnel-like structure (shown in Figure 1). This motif is characteristic of beta-propellers. Chain A serves as a peptide-recognition module. Chain Z is a short peptide derived from sclerostin (SOST). There are about four α-helices across the domain, and each acts as a connecting loop at blade interfaces (shown in Figure 2). The α-helices in 3SOV are clustered together around the major ligand-binding sites. This provides strucutral stability between the propellers and the ligand-binding-sites, and creates surface features that can be involved in interaction with other proteins. Something to note is that Chain Z is on the opposite side of the domain from the major ligand-binding sites. This may contribute the Short Linear Motif in the region which is intrinsically disordered, allowing for flexible binding and interactions. Chain Z has an unstable structure, aloowing it to serve as a docking site for other proteins and to form transient protein complexes. 
  </p>

    <br><br>Binding Site Properties
  </h4>
<p style="color: #34495e; font-family: 'Arial', sans-serif; font-size: 15px; margin-top: 18px;">
    The 3SOV beta propeller has multiple ligands: two copies of N-acetylglucosamine (NAG), four copies of Fucose (FUC), and three copies of Glycerol (GOL). Both NAG and FUC are on chain A and are N-glycans that sit on the propeller rim. N-glycans are linked to Asn residues (Asn-X-Ser/Thr sequons). These glycans stabilize the fold and influence surface properties of the β-propeller but don’t form the peptide-binding pocket, thus they are not "functional" binding sites. The glycerol is most likely an artifacts of crystallization, not physiological ligands. This is common as the crystallization buffer will land in surface cavities and hydrogen-bond weakly. 
  </p>"

The outer rim is more mixed with largely polar/charged loops (good for solubility) that have localized hydrophobic regions. N-linked glycans cover some hydrophobic spots on the rim, helping the protein stay soluble in extracellular space, and this suggests the carbohydrates coat and protect the inner hydrophobic regions. Chain Z, the sclerostin fragment, has a hydrophobic Ile side chain that neatly docks into the amphipathic pocket. The pocket is shallow and only hydrophobic in a limited section. Thats means only very short motifs with a specific makeup of polar and hydrophobic residues will fit. 
  </p>

  <h4 style="color: #7f8c8d; font-family: 'Georgia', serif; font-weight: normal; font-size: 20px; margin-top: 0;">
    β-propeller Structure
  </h4>

  <p style="color: #34495e; font-family: 'Arial', sans-serif; font-size: 15px; margin-top: 18px;">
   The crystal structure 3SOV reveals a six-bladed β-propeller domain from the human LRP6 receptor (Chain A) bound to a short sclerostin peptide fragment (Chain Z). The β-propeller is built almost entirely from β-strands and a few short α-helices in loop regions. The β-propeller displays has a hydrophobic interior and polar outer rim. N-linked glycans (NAG and FUC) line the outer rim and shield the hydrophobic inner regions, while glycerol molecules appear to be artifacs from the crystallization process. The functional binding interaction occurs in a shallow groove on the propeller’s rim, where the peptide’s conserved Asn forms hydrogen bonds with polar residues of LRP6 and the Ile side chain slots into a hydrophobic pocket. 
      <br>In contrast, the TIM barrel (3TDM) displays a different architecture with β-strands in the barrel core with surrounding α-helices. These findings sugguest β-propeller structures are highly adapted for protein-protein interactions. Their funnel-like arrangement creates surface grooves that fit short motif-specific peptide ligands. 
  </p>

   <h4 style="color: #7f8c8d; font-family: 'Georgia', serif; font-weight: normal; font-size: 20px; margin-top: 0;">
    PyMol, VMD, or ChimeraX
  </h4>

  <p style="color: #34495e; font-family: 'Arial', sans-serif; font-size: 15px; margin-top: 18px;">
    PyMOL, VMD, and ChimeraX are all powerful molecular visualization tools, but they differ in usability. PyMOL is widely used in structural biology for publication-quality figures; it emphasizes clean, customizable graphics and has an intuitive command system, though its analysis features are more limited. VMD (Visual Molecular Dynamics) certainly is a program. PyMol's interface can feel less polished for beginners. ChimeraX, the current successor to UCSF Chimera, balances high-end visualization with interactive analysis. It integrates funcitons like molecular modeling, structure alignment, and surface analysis, and, compared with PyMol, it's more user-friendly, offers a more robust command system, and provides more integration with programs like alphafold. For future use, PyMOL excels at high-quality figures, and ChimeraX provides a versatile mix of visualization and analysis in a modern, easy-to-use interface.
